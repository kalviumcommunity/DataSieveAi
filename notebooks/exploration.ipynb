# DataSieveAI - Data Exploration and Pattern Discovery
Detecting potentially harmful content using text and behavioral patterns.

```python
import sys
sys.path.append('../')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
from src.data_loader import load_liar_dataset
from src.preprocessing import TextPreprocessor
from src.feature_engineering import FeatureExtractor

sns.set(style="whitegrid")
```

## 1. Load and Clean Data

```python
df = load_liar_dataset()
print(f"Data Shape: {df.shape}")
df.head()
```

## 2. Feature Extraction

```python
extractor = FeatureExtractor()
df = extractor.extract_text_metadata(df)
df.head()
```

## 3. Label Analysis
Let's see how our target labels are distributed. In LIAR, labels range from 'false' to 'true'.

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='label', data=df, order=['pants-fire', 'false', 'barely-true', 'half-true', 'mostly-true', 'true'], palette='coolwarm')
plt.title('Distribution of Content Labels')
plt.xticks(rotation=45)
plt.show()
```

## 4. Behavioral Patterns (Simulated + Text-based)
Does misleading content spread faster or use more punctuation?

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Exclamation Marks Heatmap/Boxplot
sns.boxplot(x='label', y='exclamation_count', data=df, ax=axes[0])
axes[0].set_title('Exclamation Mark Intensity by Label')
axes[0].tick_params(axis='x', rotation=45)

# Sentiment vs Length Scatter
sns.scatterplot(x='post_length', y='sentiment_score', hue='label', alpha=0.6, data=df, ax=axes[1])
axes[1].set_title('Sentiment Score vs Post Length')

plt.tight_layout()
plt.show()
```

## 5. Visualizing Text Patterns (WordClouds)

```python
preprocessor = TextPreprocessor()
df['cleaned_statement'] = df['statement'].apply(preprocessor.clean_text)

def generate_cloud(text_data, title):
    wordcloud = WordCloud(width=800, height=400, background_color='white').generate(" ".join(text_data))
    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.title(title, fontsize=20)
    plt.axis('off')
    plt.show()

generate_cloud(df[df['label'] == 'pants-fire']['cleaned_statement'], "Common Words in Dangerous Content (Pants on Fire)")
generate_cloud(df[df['label'] == 'true']['cleaned_statement'], "Common Words in Verified Content (True)")
```
